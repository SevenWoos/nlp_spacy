## What is NLP?

* NLP is the process of getting a computer system to understand, parse, and extract human language with raw text.

* Different types of NLP:
    - Named Entity Recognition(NER)
    
    - Part of Speech Tagging(POS)
    
    - Syntatic Parsing
    
    - Text Categorization
        - read in data and categorize it.
        
        - spam detection
    
    - Coreference Resolution
    
    - Machine Translation


* NLU(Natural Language Understanding) = adjacent to NLP. Train computer systems to perform:
    - Relation Extraction
    
    - Semantic Parsing
    
    - Q&A (bots)
    
    - Summarization
    
    - Sentiment Analysis

* NLP Framework (spaCy)
    - Can use/leverage things that have been parsed syntactically/semantically (part of speech of a word, dependencies, etc).
    
    - Can leverage current Natural Language Processing methods like Transformers and Bert models.
    
    - Provides framework for doing custom training.
    
    - Scales well with large quantities of documents effectively and accurately.
